## Investigation Overview

This presentation looks at trip behavior in the Ford GoBike System, focusing mainly on trip duration and the differences between Customers and Subscribers. I focused on three main findings: most trips are relatively short, most trips are made by Subscribers, and Customers generally take longer trips than Subscribers.

## Dataset Overview and Executive Summary

The dataset contains 183,412 Ford GoBike trips recorded in the San Francisco Bay Area during February 2019. It includes information about trip duration, user type, rider characteristics, stations, and trip times.

##### Key findings:

- Most trips are relatively short, and the trip duration distribution is strongly right-skewed.
- Subscribers made the large majority of the recorded trips.
- Customers generally took longer trips than Subscribers and also showed more variation in trip duration.

### Source:
Ford GoBike System Data, provided through the Udacity Data Analyst project materials.

```python
# import all packages and set plots to be embedded inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# suppress warnings from final output
import warnings
warnings.simplefilter("ignore")
```

```python
df = pd.read_csv('201902-fordgobike-tripdata - 201902-fordgobike-tripdata.csv')

# Convert duration from seconds to minutes
df['duration_min'] = df['duration_sec'] / 60
```

## Distribution of Trip Durations

Most Ford GoBike trips were relatively short. Most trips were concentrated below about 15 minutes, with fewer trips as the duration increased. The distribution is strongly right-skewed. I limited the plot to 60 minutes so the typical trip pattern is easier to see.

```python
bins = np.arange(0, 61, 2)

plt.figure(figsize=(9, 5))
plt.hist(df['duration_min'], bins=bins)

plt.xlim(0, 60)
plt.xlabel('Trip Duration (minutes)')
plt.ylabel('Number of Trips')
plt.title('Most Ford GoBike Trips Are Short')
plt.show()
```

## Most Trips Were Made by Subscribers

Subscribers accounted for the large majority of Ford GoBike trips, while Customers represented a much smaller share. This shows that the system was used mainly by Subscribers during February 2019.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='user_type')

plt.xlabel('User Type')
plt.ylabel('Number of Trips')
plt.title('Subscribers Account for Most Ford GoBike Trips')
plt.show()
```

## Customers Tend to Take Longer Trips

Customers generally had longer trip durations than Subscribers and also showed more variation in trip length. This suggests that Customers and Subscribers may use the Ford GoBike system differently.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='user_type',
    y='duration_min'
)

plt.ylim(0, 60)
plt.xlabel('User Type')
plt.ylabel('Trip Duration (minutes)')
plt.title('Customers Tend to Take Longer Trips than Subscribers')
plt.show()
```
